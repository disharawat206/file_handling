FILE OPERATION - READ AND WRITE FILES

file handling is a crucial part of any programming language.
python provides built-in function and method to read and write to files ,text as well as binary file

```python
##reading a whole file
with open('hello.txt','r') as file:
    for r in file:
        print(r)
```

```python
##for removing newline character
with open('hello.txt','r')as file:
    for r in file:
        print(r.strip())
```

```python
##opening a file for writing (overwriting)
with open('hello.txt','w')as file:
    file.write('my age is 20 years\n')
    file.write('i am 3rd year student')##it will overwrites a file
```

```python
#write a file witout overrwriting
with open('hello.txt','a')as file:
    file.write('\nthis is a new line\n')
```

```python
##writing multiple lines
lines=['my name is disha\n','I am 20 years old']
with open('hello.txt','a') as file:
    file.writelines(lines)
```

```python
##binary files
#write a binary file
data=b'\x00\x01\x02\x03\x04\x05'
with open('example.bin','wb')as file:
     file.write(data)
```

```python
##binary files
#write a binary file
data=b'\x00\x01\x02\x03\x04\x05'
with open('example.bin','wb')as file:
     file.write(data)
```

```python
##reading binary file
with open('example.bin','rb')as file:
    data=file.read()
    print(data)
```

```python
##read content from source text file and write it to the destination text file
with open('hello.txt','r')as file1:
    content=file1.read()
with open('destination.txt','w')as file2:
    file2.write(content)    
```

```python
##read a text file and give the number of lines ,words and character
with open('hello.txt','r')as file:
    content=file.readlines()
    lines=len(content)
    words=sum(len(line.split()) for line in content)
    characters=sum(len(line) for line in content)
print(f'number of lines {lines}')
print(f'number of words {words}')
print(f'number of characters {characters}')    
```

```python
## writing and then reading a file
with open('hello.txt','w+')as file:
    file.write('new sentence\n')
    file.write('this is a new line\n')
    file.seek(0) #move the file pointer to the beginning of the file
    content=file.read()
    print(content)
```
