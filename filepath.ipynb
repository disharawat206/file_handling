```python
##joining paths
import os
dir_name='package'
file_name='hello.txt'
path=os.path.join(os.getcwd(), dir_name, file_name)
print(path)
```

```python
path='destination.txt'
if os.path.exists(path):
    print(f'{path} exists')
else:
    print(f'{path} does not exist')
        
```

```python
##checking path is file or directory
import os
path='package'
if os.path.isfile(path):
    print(f'{path} is a file')
elif os.path.isdir(path):
    print(f'{path} is a directory')
else:
    print(f'{path} does not exist')        
```

```python
##getting the absolute path of a file_path

path='example.bin'
absolute_path=os.path.abspath(path)
print(f'absolute path of {path} is {absolute_path}')
```
